## 2.2 Función de Transición

La dinámica del sistema se modela mediante una función de transición determinista:

\[
S_{t+1}=S^M(S_t,x_t,W_{t+1})
\]

donde:

- \(S_t\): estado del sistema en el periodo \(t\),
- \(x_t\): decisiones tomadas en el periodo \(t\),
- \(W_{t+1}\): información exógena observada entre \(t\) y \(t+1\).

El estado de la red evoluciona actualizando capacidades, arcos reforzados, arcos fallados, presupuesto restante y costos acumulados.

---

### Actualización de capacidades

Para cada arco \(e\):

\[
U_{e,t+1}
=
\tilde U_{e,t+1}(1+\Delta U_e r_{e,t})
\]

si el arco no falla.

En caso de falla:

\[
U_{e,t+1}
=
\tilde U_{e,t+1}
\]

donde:

- \(\tilde U_{e,t+1}\): capacidad exógena observada,
- \(r_{e,t}\in\{0,1\}\): decisión de refuerzo,
- \(\Delta U_e\): incremento porcentual de capacidad.

---

### Evolución de fallas

La condición de falla del arco evoluciona según:

\[
F_{e,t+1}
=
\xi_{e,t+1}
\]

donde:

- \(F_{e,t+1}=1\) indica arco fallado,
- \(\xi_{e,t+1}\) es el shock exógeno de falla incluido en \(W_{t+1}\).

---

### Actualización del presupuesto

El presupuesto restante evoluciona como:

\[
B_{t+1}
=
B_t
-
\sum_{e\in E} H_e r_{e,t}
-
\sum_{e\in E} R_e a_{e,t}
\]

donde:

- \(H_e\): costo de reforzar el arco,
- \(R_e\): costo de reparación,
- \(a_{e,t}\): decisión de reparación.

---

### Evolución del déficit acumulado

\[
D^{acc}_{t+1}
=
D^{acc}_t
+
\sum_{i\in N}\delta_{i,t}
\]

donde:

- \(\delta_{i,t}\): demanda no servida en el nodo \(i\).

---

### Evolución del costo acumulado

\[
C^{acc}_{t+1}
=
C^{acc}_t
+
C_t
\]

donde \(C_t\) es el costo total del periodo.

---

### Restricciones físicas

La transición respeta:

- no negatividad:

\[
U_{e,t}\ge0,
\qquad
B_t\ge0,
\qquad
\delta_{i,t}\ge0
\]

- capacidades máximas de transmisión,
- disponibilidad de generación,
- conservación física simplificada del flujo.

Toda la estocasticidad del modelo proviene exclusivamente de \(W_{t+1}\), mientras que la transición es completamente determinista dado \((S_t,x_t,W_{t+1})\).
